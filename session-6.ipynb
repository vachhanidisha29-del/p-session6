```python
import numpy as np
import pandas as pd


coaching_df = pd.read_csv("coaching_marks_clean.csv")
s
coaching_dirty = coaching_df.iloc[
    list(range(150)) + list(range(12))
].copy()

coaching_dirty.loc[
    [20, 55, 90, 130],
    "fees_paid_inr"
] = 250000

coaching_dirty
```

```python
coaching_nodup = coaching_dirty.drop_duplicates()

coaching_nodup
```

```python

Q1 = coaching_nodup["fees_paid_inr"].quantile(0.25)

Q3 = coaching_nodup["fees_paid_inr"].quantile(0.75)

IQR = Q3 - Q1

low = Q1 - (1.5 * IQR)

high = Q3 + (1.5 * IQR)


fee_outliers_iqr = coaching_nodup[
    (coaching_nodup["fees_paid_inr"] < low) |
    (coaching_nodup["fees_paid_inr"] > high)
]

fee_outliers_iqr
```

```python

fee_mean = coaching_nodup["fees_paid_inr"].mean()

fee_std = coaching_nodup["fees_paid_inr"].std()

coaching_nodup["fee_z"] = (
    coaching_nodup["fees_paid_inr"] - fee_mean
) / fee_std

fee_outliers_z = coaching_nodup[
    coaching_nodup["fee_z"].abs() > 3
]


fee_outliers_z
```
